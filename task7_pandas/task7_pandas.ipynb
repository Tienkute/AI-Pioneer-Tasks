## PHẦN 7 – PHÁT HIỆN DỮ LIỆU THIẾU (MISSING VALUES / NaN)

- **`NaN` (Not a Number)**: Đại diện cho dữ liệu bị khuyết hoặc chưa ghi nhận.
- **`df.isna()`** hoặc **`df.isnull()`**: Trả về DataFrame giá trị boolean (`True` nếu ô đó là `NaN`).
- **`df.isna().sum()`**: Đếm tổng số lượng `NaN` trên mỗi cột.
- **`df.isna().mean() * 100`**: Tính tỷ lệ phần trăm dữ liệu thiếu trên từng cột.


```python
# Thống kê chi tiết missing value cho toàn bộ dataset
missing_count = df.isna().sum()
missing_percent = (df.isna().mean() * 100).round(2)

# Tạo bảng tổng hợp các cột bị thiếu
missing_df = pd.DataFrame({
    "Số lượng Missing": missing_count,
    "Tỷ lệ Missing (%)": missing_percent
})

# Lọc các cột có thiếu dữ liệu
missing_summary = missing_df[missing_df["Số lượng Missing"] > 0]

print("=== BẢNG THỐNG KÊ CÁC CỘT CÓ DỮ LIỆU THIẾU ===")
missing_summary
```

## PHẦN 8 – XỬ LÝ DỮ LIỆU THIẾU (MISSING VALUE IMPUTATION)

Nguyên tắc vàng: **Không chỉnh sửa trực tiếp `df` gốc**, luôn tạo bản sao:
`df_clean = df.copy()`

### Các chiến lược xử lý:
1. **`dropna()` (Xóa)**:
   - Xóa hàng: Nếu dùng `df.dropna()`, dữ liệu từ 891 hàng sẽ bị giảm xuống chỉ còn 183 hàng (mất hơn 79% dữ liệu do cột Cabin thiếu quá nhiều), điều này là không hợp lý!
2. **`fillna()` (Điền dữ liệu thay thế)**:
   - **Với cột số (`Age`)**: Dùng `median()` (trung vị) thay vì `mean()` vì trung vị không bị kéo lệch bởi các giá trị ngoại lai (outliers - ví dụ những người quá già hoặc trẻ sơ sinh).
   - **Với cột phân loại (`Embarked`)**: Dùng `mode()[0]` (giá trị xuất hiện nhiều nhất - cảng Southampton 'S').
   - **Với cột thiếu quá nhiều (`Cabin`)**: Trong phạm vi bài thực hành này, ta chọn loại bỏ cột Cabin vì cột này thiếu khoảng 77% dữ liệu và không cần thiết cho các phân tích chính đang thực hiện. (Lưu ý: Trong project thực tế, việc một cột thiếu nhiều dữ liệu không tự động đồng nghĩa với việc phải xóa cột đó).


```python
# Tạo bản copy an toàn
df_clean = df.copy()

# Minh họa lý do không nên lạm dụng dropna() trên toàn bộ dataset
df_dropped_demo = df.dropna()
print("Số dòng ban đầu     :", df.shape[0])
print("Số dòng nếu dropna():", df_dropped_demo.shape[0], "-> Mất", df.shape[0] - df_dropped_demo.shape[0], "dòng (quá lãng phí dữ liệu!)")

# 1. Xử lý cột Age: Điền bằng median
age_median = df_clean["Age"].median()
df_clean["Age"] = df_clean["Age"].fillna(age_median)
print(f"\n1. Đã điền NaN cột Age bằng median = {age_median}")

# 2. Xử lý cột Embarked: Điền bằng mode
embarked_mode = df_clean["Embarked"].mode()[0]
df_clean["Embarked"] = df_clean["Embarked"].fillna(embarked_mode)
print(f"2. Đã điền NaN cột Embarked bằng mode = '{embarked_mode}'")

# 3. Xử lý cột Cabin: Bỏ cột vì thiếu > 77% dữ liệu
df_clean = df_clean.drop(columns=["Cabin"])
print("3. Đã xóa cột Cabin do tỷ lệ thiếu vượt quá 77%")

# Kiểm tra lại số lượng missing values sau xử lý
print("\n=== SỐ LƯỢNG MISSING VALUES SAU KHI LÀM SẠCH ===")
print(df_clean.isna().sum())
```

## PHẦN 9 – LỌC DỮ LIỆU BẰNG ĐIỀU KIỆN (BOOLEAN FILTERING)

Cú pháp lọc:
$$\text{df}[\text{điều kiện boolean}]$$
- Kết hợp nhiều điều kiện:
  - Phép VÀ (AND): Dùng toán tử `&` (không dùng từ khóa `and`).
  - Phép HOẶC (OR): Dùng toán tử `|` (không dùng từ khóa `or`).
- **Quy tắc bắt buộc:** Mỗi biểu thức điều kiện phải được bao trong cặp ngoặc đơn `( )`.


```python
# 1. Hành khách trên 30 tuổi
over_30 = df[df["Age"] > 30]
print("1. Số hành khách trên 30 tuổi:", len(over_30))

# 2. Hành khách là nữ (female)
females = df[df["Sex"] == "female"]
print("2. Số hành khách nữ          :", len(females))

# 3. Hành khách nữ VÀ trên 30 tuổi
females_over_30 = df[(df["Sex"] == "female") & (df["Age"] > 30)]
print("3. Số hành khách nữ trên 30 tuổi:", len(females_over_30))

# 4. Hành khách thuộc hạng vé Pclass 1
pclass_1 = df[df["Pclass"] == 1]
print("4. Số hành khách hạng Pclass 1  :", len(pclass_1))

# 5. Hành khách có giá vé (Fare) lớn hơn giá vé trung bình
fare_mean = df["Fare"].mean()
above_avg_fare = df[df["Fare"] > fare_mean]
print(f"5. Số hành khách có Fare > trung bình ({fare_mean:.2f}):", len(above_avg_fare))

print("\nVí dụ 3 dòng đầu của nhóm Fare > trung bình:")
above_avg_fare[["Name", "Pclass", "Fare"]].head(3)
```

## PHẦN 12 – GROUPBY (GOM NHÓM VÀ TÍNH TOÁN)

Quy trình hoạt động của **GroupBy**:
$$\text{Split (Chia dữ liệu)} \longrightarrow \text{Apply (Tính toán trên từng nhóm)} \longrightarrow \text{Combine (Gom kết quả lại)}$$


```python
print("=== 1. TỶ LỆ SỐNG THEO GIỚI TÍNH ===")
# Tính mean của cột Survived (vì Survived gồm 0 và 1, mean chính là tỷ lệ sống)
surv_by_sex = df.groupby("Sex")["Survived"].mean() * 100
print(surv_by_sex.round(2))

print("\n=== 2. TỶ LỆ SỐNG THEO HẠNG VÉ (Pclass) ===")
surv_by_pclass = df.groupby("Pclass")["Survived"].mean() * 100
print(surv_by_pclass.round(2))

print("\n=== 3. TUỔI TRUNG BÌNH THEO GIỚI TÍNH ===")
age_by_sex = df.groupby("Sex")["Age"].mean()
print(age_by_sex.round(2))

print("\n=== 4. GIÁ VÉ (FARE) TRUNG BÌNH THEO HẠNG VÉ ===")
fare_by_pclass = df.groupby("Pclass")["Fare"].mean()
print(fare_by_pclass.round(2))
```

## PHẦN 13 – AGG (TÍNH NHIỀU CHỈ SỐ THỐNG KÊ CÙNG LÚC)

- **`agg()`**: Cho phép linh hoạt tính toán nhiều hàm thống kê cùng một lúc trên một hoặc nhiều cột.


```python
# 1. Tính count, mean, min, max của cột Age theo Sex
print("=== 1. THỐNG KÊ TUỔI THEO GIỚI TÍNH ===")
df.groupby("Sex")["Age"].agg(["count", "mean", "min", "max"]).round(2)
```

```python
# 2. Tính các chỉ số khác nhau trên các cột khác nhau theo Pclass
print("=== 2. TỔNG HỢP NHIỀU CỘT THEO HẠNG VÉ (Pclass) ===")
df.groupby("Pclass").agg({
    "Age": ["mean", "median"],
    "Fare": ["mean", "max"],
    "Survived": "mean"
}).round(2)
```

## PHẦN 14 – THÊM / XÓA CỘT ĐƠN GIẢN

Tạo thêm các đặc trưng hữu ích từ dữ liệu có sẵn:
- **`FamilySize`**: Quy mô gia đình trên tàu = `SibSp` (anh/chị/em/vợ/chồng) + `Parch` (cha mẹ/con cái) + 1 (bản thân).

## PHẦN 19 – KIẾN THỨC CẦN NHỚ (SUMMARY CHEAT SHEET)

Bảng tra cứu ngắn gọn các hàm và khái niệm Pandas cốt lõi:

1. **`Pandas`**: Thư viện Python tiêu chuẩn chuyên xử lý, làm sạch và phân tích dữ liệu dạng bảng.
2. **`Series`**: Cấu trúc dữ liệu mảng 1 chiều có gắn nhãn index, tương ứng với một cột trong bảng.
3. **`DataFrame`**: Cấu trúc dữ liệu bảng 2 chiều gồm nhiều hàng và cột có nhãn.
4. **`read_csv`**: Hàm đọc file CSV thành DataFrame.
5. **`head`**: Xem $n$ dòng đầu tiên của bảng (mặc định 5 dòng).
6. **`tail`**: Xem $n$ dòng cuối cùng của bảng (mặc định 5 dòng).
7. **`shape`**: Thuộc tính trả về tuple `(số hàng, số cột)`.
8. **`info`**: Hiển thị tổng quan về kiểu dữ liệu từng cột, số lượng giá trị non-null và bộ nhớ sử dụng.
9. **`describe`**: Bảng thống kê tóm tắt các chỉ số quan trọng (`count`, `mean`, `std`, `min`, `max`, phân vị) của các cột số.
10. **`value_counts`**: Đếm số lần xuất hiện của các giá trị riêng biệt trong cột dạng phân loại.
11. **`isna`**: Kiểm tra dữ liệu khuyết thiếu (`NaN`), trả về boolean `True`/`False`.
12. **`dropna`**: Loại bỏ các hàng hoặc cột chứa dữ liệu thiếu.
13. **`fillna`**: Điền giá trị thay thế (như mean, median, mode) vào các ô bị `NaN`.
14. **`mean`**: Tính giá trị trung bình cộng của dữ liệu dạng số.
15. **`median`**: Tính giá trị trung vị (không bị ảnh hưởng bởi giá trị ngoại lai).
16. **`mode`**: Tìm giá trị xuất hiện nhiều nhất trong cột (thích hợp cho dữ liệu phân loại).
17. **`loc`**: Truy xuất dữ liệu theo nhãn tên của hàng và cột (`df.loc[row_label, col_label]`).
18. **`iloc`**: Truy xuất dữ liệu theo chỉ số vị trí số nguyên (`df.iloc[row_idx, col_idx]`).
19. **`filtering`**: Kỹ thuật lọc hàng thỏa mãn điều kiện boolean với cú pháp `df[điều kiện]`.
20. **`sort_values`**: Sắp xếp bảng theo giá trị của một hoặc nhiều cột tăng dần/giảm dần.
21. **`groupby`**: Chia dữ liệu thành các nhóm theo cột để áp dụng hàm tính toán (*Split-Apply-Combine*).
22. **`agg`**: Cho phép tính toán linh hoạt nhiều chỉ số thống kê cùng lúc trên các cột.
23. **`duplicated`**: Phát hiện và đếm các dòng dữ liệu bị trùng lặp hoàn toàn.
