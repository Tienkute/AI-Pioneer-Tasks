## Phần 7 – Phát hiện dữ liệu thiếu (Missing Values)

`NaN` là ký hiệu cho ô dữ liệu bị thiếu. Cách mình kiểm tra:
- `df.isna().sum()`: đếm số ô NaN trên từng cột.

```python
# Thống kê chi tiết missing value cho toàn bộ dataset
missing_count = df.isna().sum()
missing_percent = (df.isna().mean() * 100).round(2)

# Tạo bảng tổng hợp các cột bị thiếu
missing_df = pd.DataFrame({
    "Số lượng Missing": missing_count,
    "Tỷ lệ Missing (%)": missing_percent
})

# Lọc các cột có thiếu dữ liệu
missing_summary = missing_df[missing_df["Số lượng Missing"] > 0]

print("=== BẢNG THỐNG KÊ CÁC CỘT CÓ DỮ LIỆU THIẾU ===")
missing_summary
```

## Phần 8 – Xử lý dữ liệu thiếu

Mình copy sang `df_clean = df.copy()` trước khi làm gì – để giữ nguyên bản gốc, tránh bị đè dữ liệu.

Với từng cột trong Titanic, mình chọn cách xử lý theo lý do cụ thể:
- **Cột `Age`** (thiếu 177 ô): điền bằng **`median`** thay vì `mean`, vì median không bị kéo lệch bởi outlier (trẻ sơ sinh hoặc người rất già).
- **Cột `Embarked`** (thiếu 2 ô): điền bằng **`mode`** – giá trị xuất hiện nhiều nhất trong cột phân loại, ở đây là cảng `'S'` (Southampton).
- **Cột `Cabin`** (thiếu 687 ô, ~77%): mình drop cột này vì thiếu quá nhiều và không dùng cho các bước phân tích sau.


```python
# Tạo bản copy an toàn
df_clean = df.copy()

# Minh họa lý do không nên lạm dụng dropna() trên toàn bộ dataset
df_dropped_demo = df.dropna()
print("Số dòng ban đầu     :", df.shape[0])
print("Số dòng nếu dropna():", df_dropped_demo.shape[0], "-> Mất", df.shape[0] - df_dropped_demo.shape[0], "dòng (quá lãng phí dữ liệu!)")

# 1. Xử lý cột Age: Điền bằng median
age_median = df_clean["Age"].median()
df_clean["Age"] = df_clean["Age"].fillna(age_median)
print(f"\n1. Đã điền NaN cột Age bằng median = {age_median}")

# 2. Xử lý cột Embarked: Điền bằng mode
embarked_mode = df_clean["Embarked"].mode()[0]
df_clean["Embarked"] = df_clean["Embarked"].fillna(embarked_mode)
print(f"2. Đã điền NaN cột Embarked bằng mode = '{embarked_mode}'")

# 3. Xử lý cột Cabin: Bỏ cột vì thiếu > 77% dữ liệu
df_clean = df_clean.drop(columns=["Cabin"])
print("3. Đã xóa cột Cabin do tỷ lệ thiếu vượt quá 77%")

# Kiểm tra lại số lượng missing values sau xử lý
print("\n=== SỐ LƯỢNG MISSING VALUES SAU KHI LÀM SẠCH ===")
print(df_clean.isna().sum())
```

## Phần 9 – Lọc dữ liệu bằng điều kiện (Boolean Filtering)

Cú pháp: `df[điều kiện boolean]`. Khi kết hợp nhiều điều kiện:
- Phép VÀ: dùng `&` (không dùng từ khóa `and`).
- Phép HOẶC: dùng `|` (không dùng từ khóa `or`).
- Mỗi điều kiện phải bọc trong cặp ngoặc đơn `()`.


```python
# 1. Hành khách trên 30 tuổi
over_30 = df[df["Age"] > 30]
print("1. Số hành khách trên 30 tuổi:", len(over_30))

# 2. Hành khách là nữ (female)
females = df[df["Sex"] == "female"]
print("2. Số hành khách nữ          :", len(females))

# 3. Hành khách nữ VÀ trên 30 tuổi
females_over_30 = df[(df["Sex"] == "female") & (df["Age"] > 30)]
print("3. Số hành khách nữ trên 30 tuổi:", len(females_over_30))

# 4. Hành khách thuộc hạng vé Pclass 1
pclass_1 = df[df["Pclass"] == 1]
print("4. Số hành khách hạng Pclass 1  :", len(pclass_1))

# 5. Hành khách có giá vé (Fare) lớn hơn giá vé trung bình
fare_mean = df["Fare"].mean()
above_avg_fare = df[df["Fare"] > fare_mean]
print(f"5. Số hành khách có Fare > trung bình ({fare_mean:.2f}):", len(above_avg_fare))

print("\nVí dụ 3 dòng đầu của nhóm Fare > trung bình:")
above_avg_fare[["Name", "Pclass", "Fare"]].head(3)
```

## Phần 12 – Gom nhóm dữ liệu với groupby()

Mình dùng `groupby` để chia dữ liệu thành từng nhóm rồi tính thống kê cho mỗi nhóm. Cơ chế hoạt động theo 3 bước: **Split** (chia nhóm) → **Apply** (tính toán) → **Combine** (ghép kết quả lại).

Trong Titanic mình dùng phần này để xem tỷ lệ sống sót theo giới tính và Pclass.


```python
print("=== 1. TỶ LỆ SỐNG THEO GIỚI TÍNH ===")
# Tính mean của cột Survived (vì Survived gồm 0 và 1, mean chính là tỷ lệ sống)
surv_by_sex = df.groupby("Sex")["Survived"].mean() * 100
print(surv_by_sex.round(2))

print("\n=== 2. TỶ LỆ SỐNG THEO HẠNG VÉ (Pclass) ===")
surv_by_pclass = df.groupby("Pclass")["Survived"].mean() * 100
print(surv_by_pclass.round(2))

print("\n=== 3. TUỔI TRUNG BÌNH THEO GIỚI TÍNH ===")
age_by_sex = df.groupby("Sex")["Age"].mean()
print(age_by_sex.round(2))

print("\n=== 4. GIÁ VÉ (FARE) TRUNG BÌNH THEO HẠNG VÉ ===")
fare_by_pclass = df.groupby("Pclass")["Fare"].mean()
print(fare_by_pclass.round(2))
```

## Phần 13 – Tính nhiều chỉ số với agg()

`agg()` cho phép tính nhiều hàm thống kê cùng lúc trên một hoặc nhiều cột. Ví dụ: tính cả `mean` tỷ lệ sống, `median` tuổi và `count` số hành khách theo Pclass chỉ trong một lệnh.


```python
# 1. Tính count, mean, min, max của cột Age theo Sex
print("=== 1. THỐNG KÊ TUỔI THEO GIỚI TÍNH ===")
df.groupby("Sex")["Age"].agg(["count", "mean", "min", "max"]).round(2)
```

```python
# 2. Tính các chỉ số khác nhau trên các cột khác nhau theo Pclass
print("=== 2. TỔNG HỢP NHIỀU CỘT THEO HẠNG VÉ (Pclass) ===")
df.groupby("Pclass").agg({
    "Age": ["mean", "median"],
    "Fare": ["mean", "max"],
    "Survived": "mean"
}).round(2)
```

## Phần 14 – Tạo thêm cột mới

Tạo thêm các đặc trưng từ dữ liệu có sẵn:
- `FamilySize`: `SibSp + Parch + 1` (tổng số thành viên gia đình trên tàu, kể cả bản thân).

## Phần 16 – Mini Project: Khai phá dữ liệu (EDA) Titanic

Áp dụng tất cả các bước đã học vào dataset Titanic – từ load dữ liệu, làm sạch missing values, đến groupby và thống kê để rút ra nhận xét từ số liệu thực tế.


```python
# 1. Load dataset
df_eda = pd.read_csv(csv_path)

# 2. Hiển thị shape
print("1 & 2. Shape ban đầu của dataset:", df_eda.shape)

# 3. Kiểm tra missing values
print("\n3. Missing values ban đầu:\n", df_eda.isna().sum())

# 4. Xử lý Age bằng median
df_eda["Age"] = df_eda["Age"].fillna(df_eda["Age"].median())

# 5. Xử lý Embarked bằng mode
df_eda["Embarked"] = df_eda["Embarked"].fillna(df_eda["Embarked"].mode()[0])

# 6. Bỏ Cabin do thiếu > 77%
df_eda = df_eda.drop(columns=["Cabin"])
print("\n4, 5, 6. Đã xử lý xong dữ liệu thiếu! Missing values sau xử lý:\n", df_eda.isna().sum())
```

```python
# 7. Tính số hành khách nam/nữ
print("7. Số hành khách nam/nữ:\n", df_eda["Sex"].value_counts())

# 8. Tính tỷ lệ sống toàn bộ
overall_survival_rate = df_eda["Survived"].mean()
print(f"\n8. Tỷ lệ sống sót toàn bộ dataset: {overall_survival_rate * 100:.2f}%")

# 9. Tính tỷ lệ sống theo Sex
print("\n9. Tỷ lệ sống theo Sex (%):\n", (df_eda.groupby("Sex")["Survived"].mean() * 100).round(2))

# 10. Tính tỷ lệ sống theo Pclass
print("\n10. Tỷ lệ sống theo Pclass (%):\n", (df_eda.groupby("Pclass")["Survived"].mean() * 100).round(2))

# 11 & 12. Tuổi trung bình & Fare trung bình
print(f"\n11. Tuổi trung bình : {df_eda['Age'].mean():.2f}")
print(f"12. Fare trung bình : {df_eda['Fare'].mean():.2f}")
```

```python
# 13. Tìm hành khách có Fare cao nhất
max_fare_passenger = df_eda.loc[df_eda["Fare"].idxmax()]
print(f"13. Hành khách có Fare cao nhất: {max_fare_passenger['Name']} (Fare: {max_fare_passenger['Fare']})")

# 14. Tìm hành khách lớn tuổi nhất
max_age_passenger = df_eda.loc[df_eda["Age"].idxmax()]
print(f"14. Hành khách lớn tuổi nhất: {max_age_passenger['Name']} (Tuổi: {max_age_passenger['Age']})")

# 15 & 16. Tạo FamilySize và tính trung bình
df_eda["FamilySize"] = df_eda["SibSp"] + df_eda["Parch"] + 1
print(f"\n15 & 16. FamilySize trung bình: {df_eda['FamilySize'].mean():.2f}")

# 17. Sắp xếp top 10 hành khách có Fare cao nhất
print("\n17. Top 10 hành khách có Fare cao nhất:")
df_eda.sort_values("Fare", ascending=False)[["Name", "Pclass", "Age", "Fare", "Survived"]].head(10)
```

```python
# 18. Tạo bảng groupby tổng hợp chi tiết theo Sex và Pclass
print("18. Bảng groupby tổng hợp theo Sex và Pclass:")
summary = df_eda.groupby(["Sex", "Pclass"]).agg(
    passenger_count=("PassengerId", "count"),
    average_age=("Age", "mean"),
    average_fare=("Fare", "mean"),
    survival_rate=("Survived", "mean")
).round(2)

summary
```

## Phần 17 – Nhận xét dựa trên kết quả thực hành

Những gì mình quan sát được từ dataset Titanic sau khi chạy các bước phân tích:

### Code giải bài tập


```python
print("=== ĐÁP ÁN BÀI 1: ĐẾM SỐ HÀNH KHÁCH DƯỚI 18 TUỔI ===")
under_18_count = (df_clean["Age"] < 18).sum()
print("Số hành khách dưới 18 tuổi:", under_18_count)

print("\n=== ĐÁP ÁN BÀI 2: FARE TRUNG BÌNH CỦA NỮ Ở PCLASS 1 ===")
female_pclass1_fare = df_clean[(df_clean["Sex"] == "female") & (df_clean["Pclass"] == 1)]["Fare"].mean()
print(f"Giá vé trung bình của nữ hạng 1: ${female_pclass1_fare:.2f}")

print("\n=== ĐÁP ÁN BÀI 3: PCLASS CÓ TUỔI TRUNG BÌNH CAO NHẤT ===")
age_by_pclass = df_clean.groupby("Pclass")["Age"].mean()
highest_age_pclass = age_by_pclass.idxmax()
print(age_by_pclass.round(2))
print(f"-> Hạng vé có tuổi trung bình cao nhất là Pclass: {highest_age_pclass}")

print("\n=== ĐÁP ÁN BÀI 4: 5 HÀNH KHÁCH LỚN TUỔI NHẤT ===")
top5_oldest = df_clean.sort_values("Age", ascending=False)[["Name", "Sex", "Age", "Pclass"]].head(5)
display(top5_oldest)

print("\n=== ĐÁP ÁN BÀI 5: TỶ LỆ SỐNG CỦA HÀNH KHÁCH DƯỚI 18 TUỔI ===")
under_18_survival = df_clean[df_clean["Age"] < 18]["Survived"].mean() * 100
print(f"Tỷ lệ sống của hành khách < 18 tuổi: {under_18_survival:.2f}%")

print("\n=== ĐÁP ÁN BÀI 6: GROUPBY SEX VÀ PCLASS ĐỂ ĐẾM SỐ HÀNH KHÁCH ===")
count_by_sex_pclass = df_clean.groupby(["Sex", "Pclass"])["PassengerId"].count()
print(count_by_sex_pclass)

print("\n=== ĐÁP ÁN BÀI 7: TẠO CỘT PHÂN LOẠI NHÓM TUỔI (AgeGroup) ===")
bins = [0, 12, 18, 60, 120]
labels = ["Child", "Teen", "Adult", "Senior"]
df_clean["AgeGroup"] = pd.cut(df_clean["Age"], bins=bins, labels=labels, right=False)
print("Số lượng hành khách theo từng nhóm tuổi:")
print(df_clean["AgeGroup"].value_counts())
display(df_clean[["Name", "Age", "AgeGroup"]].head(5))
```
